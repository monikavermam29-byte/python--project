```python
def remove_char(x):
    kmpl, cc, bhp = x
    kmpl = kmpl.split(' ')[0]
    cc = cc.split(' ')[0]
    bhp = bhp.split(' ')[0]

    return kmpl, cc, bhp
```

```python
cars[['Mileage(KMPL)', 'Engine(CC)', 'Max_power(bhp)']] = cars[['mileage', 'engine', 'max_power']].apply(remove_char, axis = 1, result_type = 'expand')
```
